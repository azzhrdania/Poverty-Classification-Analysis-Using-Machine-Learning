9. Cek Outlier

```python
numeric_cols = ['Kemiskinan','RLS','Pengeluaran','IPM',
                'UHH','Sanitasi','AirMinum','TPT','TPAK']

for col in numeric_cols:
  Q1 = df[col].quantile(0.25)
  Q3 = df[col].quantile(0.75)
  IQR = Q3 - Q1

  lower = Q1 - 1.5 * IQR
  upper = Q3 + 1.5 * IQR

  outlier_count = ((df[col] < lower) | (df[col] > upper)).sum()

  print(f"{col}: {outlier_count} outlier")
```

```python
numeric_cols = ['Kemiskinan','RLS','Pengeluaran','IPM', 'UHH','Sanitasi','AirMinum','TPT','TPAK']

plt.figure(figsize=(15,10))

for i, col in enumerate(numeric_cols, 1):
    plt.subplot(3,3,i)
    sns.boxplot(x=df[col])
    plt.title(col)

plt.tight_layout()
plt.show()
```

# Data Cleaning

1. penanganan missing value

```python
df.dropna(inplace=True)
```

2. validasi penanganan missing value

```python
df.isnull().sum()
```

3. heatmap setelah cleaning

```python
sns.heatmap(df.isnull(), cbar=False)
```

4. penanganan outlier

```python
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df = df[(df[col] >= lower) & (df[col] <= upper)]

    print("Outlier berhasil dihapus")
```

5. validasi penanganan outlier

```python
plt.figure(figsize=(15,10))

for i, col in enumerate(numeric_cols, 1):
    plt.subplot(3,3,i)
    sns.boxplot(x=df[col])
    plt.title(col)

plt.tight_layout()
plt.show()
```

6. penanganan data duplikat

```python
df.drop_duplicates(inplace=True)
```

7. validasi penanganan data duplikat

```python
df.duplicated().sum()
```

# EDA

```python
plt.figure(figsize=(10,8))

corr = df.select_dtypes(include='number').corr()

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Korelasi Antar Variabel')

plt.show()
```

Berdasarkan heatmap korelasi, variabel seperti IPM, Pengeluaran, RLS, dan UHH umumnya memiliki hubungan negatif terhadap tingkat kemiskinan. Artinya, semakin tinggi kualitas pendidikan, kesehatan, dan kesejahteraan masyarakat, maka tingkat kemiskinan cenderung menurun. Informasi ini dapat digunakan sebagai dasar dalam menentukan prioritas program pembangunan daerah.

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df['Kemiskinan'],
    bins=15,
    kde=True
)

plt.title('Distribusi Tingkat Kemiskinan')
plt.xlabel('Persentase Kemiskinan')
plt.ylabel('Frekuensi')

plt.show()
```

Sebagian besar wilayah memiliki tingkat kemiskinan pada rentang tertentu, namun terdapat beberapa wilayah dengan tingkat kemiskinan yang jauh lebih tinggi dibandingkan wilayah lainnya. Hal ini menunjukkan adanya ketimpangan kondisi ekonomi antar daerah yang perlu menjadi perhatian dalam perumusan kebijakan.

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x='Pengeluaran',
    y='Kemiskinan'
)

plt.title('Hubungan Pengeluaran dan Kemiskinan')
plt.xlabel('Pengeluaran')
plt.ylabel('Kemiskinan')

plt.show()
```

Terlihat kecenderungan hubungan negatif antara pengeluaran dan tingkat kemiskinan. Wilayah dengan tingkat pengeluaran yang lebih tinggi cenderung memiliki tingkat kemiskinan yang lebih rendah. Temuan ini menunjukkan bahwa peningkatan daya beli masyarakat dapat berkontribusi pada penurunan tingkat kemiskinan.

```python
top5 = df.nlargest(10, 'Kemiskinan')

plt.figure(figsize=(10,6))

sns.barplot(
    data=top5,
    x='Kemiskinan',
    y='Provinsi'
)

plt.title('5 Wilayah dengan Tingkat Kemiskinan Tertinggi')
plt.xlabel('Kemiskinan (%)')
plt.ylabel('Provinsi')

plt.show()
```

Visualisasi menunjukkan wilayah yang memiliki tingkat kemiskinan tertinggi dibandingkan wilayah lainnya. Informasi ini dapat digunakan sebagai dasar penentuan prioritas program bantuan sosial dan pembangunan ekonomi yang lebih terfokus pada daerah yang membutuhkan.

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x='IPM',
    y='Pengeluaran'
)

plt.title('Hubungan IPM dan Pengeluaran')
plt.xlabel('Indeks Pembangunan Manusia (IPM)')
plt.ylabel('Pengeluaran')

plt.show()
```

Wilayah dengan nilai IPM yang lebih tinggi umumnya memiliki tingkat pengeluaran yang juga tinggi. Temuan ini menunjukkan bahwa pembangunan kualitas manusia dan pengeluaran memiliki korelasi yang saling berhubungan antar fiturnya.
